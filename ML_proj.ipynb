```python
import pandas as pd
import numpy as np
```

#### loading dataset

```python
df = pd.read_csv('crop_production.csv')
```

```python
df.shape
```

```python
df.head()
```

```python
df.isnull().sum() #missing values check 
```

```python
df = df.dropna(subset=['Production']) #cleaning
```

```python
df = df[df['Area'] > 0]
```

```python
df['Yield'] = df['Production'] / df['Area']
```

```python
categorical_cols = ['State_Name', 'District_Name', 'Season', 'Crop']
for col in categorical_cols:
    df[col] = df[col].astype(str).str.strip().str.lower()
```

#### Train Test Split

```python
df_train = df[df['Crop_Year'] < 2014].copy()
df_test = df[df['Crop_Year'] >= 2014].copy()

print(f"Training data size: {df_train.shape[0]} rows")
print(f"Testing data size: {df_test.shape[0]} rows")
```

```python
features = ['State_Name', 'District_Name', 'Season', 'Crop', 'Crop_Year']
target = 'Yield'
```

```python
X_train = df_train[features]
y_train = df_train[target]
```

```python
X_test = df_test[features]
y_test = df_test[target]
```
